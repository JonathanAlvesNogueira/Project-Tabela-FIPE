```python
lista = [
    # 'janeiro/2016', 'fevereiro/2016', 'março/2016', 'abril/2016', 'maio/2016',
    # 'junho/2016', 'julho/2016', 'agosto/2016', 'setembro/2016', 'outubro/2016',
    # 'novembro/2016', 'dezembro/2016',
    # 'janeiro/2017', 'fevereiro/2017', 'março/2017', 'abril/2017', 'maio/2017',
    # 'junho/2017', 'julho/2017', 'agosto/2017', 'setembro/2017', 'outubro/2017',
    # 'novembro/2017', 'dezembro/2017',
    # 'janeiro/2018', 'fevereiro/2018', 'março/2018', 'abril/2018', 'maio/2018',
    # 'junho/2018', 'julho/2018', 'agosto/2018', 'setembro/2018', 'outubro/2018',
    # 'novembro/2018', 'dezembro/2018',
    # 'janeiro/2019', 'fevereiro/2019', 'março/2019', 'abril/2019', 'maio/2019',
    # 'junho/2019', 'julho/2019', 'agosto/2019', 'setembro/2019', 'outubro/2019',
    # 'novembro/2019', 'dezembro/2019',
    # 'janeiro/2020', 'fevereiro/2020', 'março/2020', 'abril/2020', 'maio/2020',
    # 'junho/2020', 'julho/2020', 'agosto/2020', 'setembro/2020', 'outubro/2020',
    # 'novembro/2020', 'dezembro/2020',
    # 'janeiro/2021', 'fevereiro/2021', 'março/2021', 'abril/2021', 'maio/2021',
    # 'junho/2021', 'julho/2021', 'agosto/2021', 'setembro/2021', 'outubro/2021',
    # 'novembro/2021', 'dezembro/2021',
    'janeiro/2022', 'fevereiro/2022', 'março/2022', 'abril/2022', 'maio/2022',
    'junho/2022', 'julho/2022', 'agosto/2022', 'setembro/2022', 'outubro/2022',
    'novembro/2022', 'dezembro/2022',
    'janeiro/2023', 'fevereiro/2023', 'março/2023', 'abril/2023', 'maio/2023',
    'junho/2023', 'julho/2023', 'agosto/2023', 'setembro/2023', 'outubro/2023',
    'novembro/2023', 'dezembro/2023',
    'janeiro/2024', 'fevereiro/2024', 'março/2024', 'abril/2024', 'maio/2024',
    'junho/2024', 'julho/2024', 'agosto/2024', 'setembro/2024', 'outubro/2024',
    'novembro/2024', 'dezembro/2024',
    'janeiro/2025', 'fevereiro/2025', 'março/2025', 'abril/2025', 'maio/2025',
    'junho/2025', 'julho/2025', 'agosto/2025', 'setembro/2025', 'outubro/2025',
    'novembro/2025', 'dezembro/2025',
    'janeiro/2026', 'fevereiro/2026', 'março/2026', 'abril/2026', 'maio/2026',
    'junho/2026', 'julho/2026', 'agosto/2026',
]

# ============================================================
# FUNÇÃO PARA SELECIONAR MÊS DE REFERÊNCIA
# ============================================================

def selecionar_mes(mes):

    campo = wait.until(
        EC.element_to_be_clickable(
            (
                By.XPATH,
                "//select[@id='selectTabelaReferenciacarro']"
                "/following-sibling::div[contains(@class,'chosen-container')]"
            )
        )
    )

    campo.click()

    time.sleep(0.5)

    opcoes = wait.until(
        EC.presence_of_all_elements_located(
            (By.CSS_SELECTOR, "li.active-result")
        )
    )

    for opcao in opcoes:

        if opcao.text.strip() == mes:

            opcao.click()

            time.sleep(1)

            return

    raise Exception(f"Mês não encontrado: {mes}")


# ============================================================
# FUNÇÃO PARA SELECIONAR CARRO
# ============================================================

def selecionar_carro(marca, modelo, ano):

    # --------------------------------------------------------
    # MARCA
    # --------------------------------------------------------

    campo = wait.until(
        EC.element_to_be_clickable(
            (
                By.XPATH,
                "//select[@id='selectMarcacarro']"
                "/following-sibling::div[contains(@class,'chosen-container')]"
            )
        )
    )

    campo.click()

    time.sleep(0.5)

    opcoes = wait.until(
        EC.presence_of_all_elements_located(
            (By.CSS_SELECTOR, "li.active-result")
        )
    )

    for opcao in opcoes:

        if opcao.text.strip() == marca:

            opcao.click()

            break

    else:

        raise Exception(f"Marca não encontrada: {marca}")


    # --------------------------------------------------------
    # MODELO
    # --------------------------------------------------------

    campo = wait.until(
        EC.element_to_be_clickable(
            (
                By.XPATH,
                "//select[@id='selectAnoModelocarro']"
                "/following-sibling::div[contains(@class,'chosen-container')]"
            )
        )
    )

    campo.click()

    time.sleep(0.5)

    opcoes = wait.until(
        EC.presence_of_all_elements_located(
            (By.CSS_SELECTOR, "li.active-result")
        )
    )

    for opcao in opcoes:
        if opcao.text.strip() == modelo:
            opcao.click()
            break
    else:
        raise Exception(f"Modelo não encontrado: {modelo}")


    # --------------------------------------------------------
    # ANO
    # --------------------------------------------------------
    campo = wait.until(
        EC.element_to_be_clickable(
            (
                By.XPATH,
                "//select[@id='selectAnocarro']"
                "/following-sibling::div[contains(@class,'chosen-container')]"
            )
        )
    )
    campo.click()
    time.sleep(0.5)
    opcoes = wait.until(
        EC.presence_of_all_elements_located(
            (By.CSS_SELECTOR, "li.active-result")
        )
    )
    for opcao in opcoes:
        if opcao.text.strip() == ano:
            opcao.click()
            break
    else:

        raise Exception(f"Ano não encontrado: {ano}")


    # --------------------------------------------------------
    # RESULTADO
    # --------------------------------------------------------

    print("Marca :", marca)
    print("Modelo:", modelo)
    print("Ano   :", ano)


# ============================================================
# FUNÇÃO PARA EXTRAIR RESULTADO FIPE
# ============================================================

def extrair_fipe():

    tabela = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, "//tbody[.//td[contains(., 'Preço Médio')]]")
        )
    )

    dados = {}

    for linha in tabela.find_elements(By.TAG_NAME, "tr"):
        colunas = linha.find_elements(By.TAG_NAME, "td")

        if len(colunas) == 2:
            campo = colunas[0].text.strip().replace(":", "")
            valor = colunas[1].text.strip()
            dados[campo] = valor

    return dados


# ============================================================
# LOOP: PARA CADA CARRO DO SQL -> TODOS OS MESES
# ============================================================

# ============================================================
# GUARDIÃO: combinações (veículo, mês) já coletadas na tabela
# ============================================================

from sqlalchemy import text

query_preenchidos = text(
    "SELECT `Marca`, `Modelo`, `Ano Modelo`, `Mês` FROM valores_fipe"
)

with engine.connect() as conn:
    preenchidos = {
        (r[0], r[1], r[2], r[3])
        for r in conn.execute(query_preenchidos)
    }

print(f"Combinações já coletadas na tabela: {len(preenchidos)}")

resultados = []

for _, linha in df.iterrows():

    marca = linha["Marca"]
    modelo = linha["Modelo"]
    ano = linha["Ano"]

    print("\n" + "=" * 60)
    print(f"CARRO: {marca} | {modelo} | {ano}")
    print("=" * 60)

    for mes in lista:

        if (marca, modelo, ano, mes) in preenchidos:

            print(f"Já coletado - {mes} (pulando)")

            continue

        try:

            selecionar_mes(mes)

            selecionar_carro(marca, modelo, ano)

            botao = wait.until(
                EC.element_to_be_clickable(
                    (By.XPATH, "//a[normalize-space()='Pesquisar']")
                )
            )

            botao.click()

            time.sleep(3)

            dados = extrair_fipe()

            dados["Mês"] = mes
            dados["Marca"] = marca
            dados["Modelo"] = modelo

            resultados.append(dados)

            from sqlalchemy import text

            insert_sql = text("""
                INSERT IGNORE INTO valores_fipe
                    (`Mês de referência`, `Código Fipe`, `Marca`, `Modelo`,
                     `Ano Modelo`, `Autenticação`, `Data da consulta`,
                     `Preço Médio`, `Mês`)
                VALUES
                    (:mes_referencia, :codigo_fipe, :marca, :modelo,
                     :ano_modelo, :autenticacao, :data_consulta,
                     :preco_medio, :mes)
            """)

            with engine.begin() as conn:
                conn.execute(insert_sql, {
                    "mes_referencia": dados["Mês de referência"],
                    "codigo_fipe": dados["Código Fipe"],
                    "marca": dados["Marca"],
                    "modelo": dados["Modelo"],
                    "ano_modelo": dados["Ano Modelo"],
                    "autenticacao": dados.get("Autenticação"),
                    "data_consulta": dados.get("Data da consulta"),
                    "preco_medio": dados.get("Preço Médio"),
                    "mes": mes,
                })
            preenchidos.add((marca, modelo, ano, mes))

            print(f"OK - {mes}")
        except Exception as e:
            print(f"Erro - {mes}: {e}")
df_resultado = pd.DataFrame(resultados)
df_resultado
```
