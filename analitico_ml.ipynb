Carrega os CSVs de **`dados_ja_extraidos/`** (extrações mensais de janeiro a agosto/2026) e prepara um DataFrame único para análise e modelagem.

```python
import pandas as pd
import os

PASTA_CSVS = r"C:\Users\joth1\Desktop\Project FIPE - DataScience\Project-Tabela-FIPE\dados_ja_extraidos"
```

```python
arquivos = sorted(f for f in os.listdir(PASTA_CSVS) if f.endswith(".csv"))
print(f"Arquivos CSV encontrados: {len(arquivos)}")
arquivos
```

## 1. Leitura dos CSVs

**Delimitador verificado:** os arquivos usam **TAB** (`\t`) como separador, e não espaço. Encoding **UTF-8**. Todos têm as mesmas 12 colunas:
`tipo_veiculo`, `codigo_fipe`, `nome_modelo`, `nome_marca`, `nome_combustivel`, `sigla_combustivel`, `ano_modelo`, `zero_km`, `valor_centavos`, `valor_formatado`, `mes_referencia`, `ano_referencia`.

```python
frames = {}
for f in arquivos:
    caminho = os.path.join(PASTA_CSVS, f)
    frames[f] = pd.read_csv(caminho, sep="\t", encoding="utf-8")
    print(f"{f}: {len(frames[f]):,} linhas")
```

## 2. Consolidação
Une todos os meses em um único DataFrame para análise.

```python
df_fipe = pd.concat(frames.values(), ignore_index=True)
print(f"Total de registros: {len(df_fipe):,}")
# df_fipe.head()
df_fipe
```
