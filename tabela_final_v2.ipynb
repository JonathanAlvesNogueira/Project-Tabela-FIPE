```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import (
    ElementClickInterceptedException,
    StaleElementReferenceException,
    TimeoutException
)
import pandas as pd
import time
lista = []
# ============================================================
# CONFIGURAÇÃO
# ============================================================

driver = webdriver.Chrome()

wait = WebDriverWait(driver, 20)

driver.get("https://veiculos.fipe.org.br/")

driver.maximize_window()

time.sleep(3)


marca_solicitada = "Fiat"



# ============================================================
# 1. ABRIR "CONSULTA DE CARROS E UTILITÁRIOS PEQUENOS"
# ============================================================

titulo = wait.until(
    EC.presence_of_element_located(
        (
            By.XPATH,
            "//div[contains(@class,'title') and "
            "normalize-space()='Consulta de Carros e Utilitários Pequenos']"
        )
    )
)

# Leva o elemento para uma posição visível
driver.execute_script(
    "arguments[0].scrollIntoView({block: 'center'});",
    titulo
)

time.sleep(0.5)

# Clique via JavaScript para evitar problemas de overlay
driver.execute_script(
    "arguments[0].click();",
    titulo
)

time.sleep(2)


# ============================================================
# IDS DOS SELECTS
# ============================================================

SELECT_MARCA = "selectMarcacarro"
SELECT_MODELO = "selectAnoModelocarro"
SELECT_ANO = "selectAnocarro"


# ============================================================
# FUNÇÃO: ABRIR SELECT CHOSEN
# ============================================================

def abrir_select(id_select, marca_solicitada):

    xpath = (
        f"//select[@id='{id_select}']"
        f"/following-sibling::div[contains(@class,'chosen-container')]"
    )

    campo = wait.until(
        EC.presence_of_element_located(
            (By.XPATH, xpath)
        )
    )

    # Coloca o elemento no centro da tela
    driver.execute_script(
        "arguments[0].scrollIntoView({block: 'center'});",
        campo
    )

    time.sleep(0.3)

    # Tenta clique normal
    try:

        campo.click()

    except ElementClickInterceptedException:

        # Se houver algum elemento sobrepondo,
        # usa JavaScript
        driver.execute_script(
            "arguments[0].click();",
            campo
        )

    time.sleep(0.5)

    return campo


# ============================================================
# FUNÇÃO: PEGAR OPÇÕES DO DROPDOWN
# ============================================================

def pegar_opcoes():

    wait.until(
        EC.presence_of_all_elements_located(
            (
                By.CSS_SELECTOR,
                "li.active-result"
            )
        )
    )

    opcoes = driver.find_elements(
        By.CSS_SELECTOR,
        "li.active-result"
    )

    lista = []

    for opcao in opcoes:

        try:

            texto = opcao.text.strip()

            if texto:
                lista.append(texto)

        except StaleElementReferenceException:

            continue

    return lista


# ============================================================
# FUNÇÃO: CLICAR EM UMA OPÇÃO DO CHOSEN
# ============================================================

def clicar_opcao(texto):

    opcoes = driver.find_elements(
        By.CSS_SELECTOR,
        "li.active-result"
    )

    for opcao in opcoes:

        try:

            texto_opcao = opcao.text.strip()

            if texto_opcao == texto:

                # Coloca a opção no centro da tela
                driver.execute_script(
                    "arguments[0].scrollIntoView({block: 'center'});",
                    opcao
                )

                time.sleep(0.2)

                try:

                    # Primeiro tenta clique normal
                    opcao.click()

                except ElementClickInterceptedException:

                    # Se houver overlay, usa JS
                    driver.execute_script(
                        "arguments[0].click();",
                        opcao
                    )

                time.sleep(0.8)

                return True

        except StaleElementReferenceException:

            continue

    return False


# ============================================================
# 2. ABRIR SELECT DE MARCAS
# ============================================================

abrir_select(SELECT_MARCA, marca_solicitada)


# ============================================================
# 3. PEGAR TODAS AS MARCAS
# ============================================================

marcas = pegar_opcoes()

print()
print("=" * 70)
print("MARCAS ENCONTRADAS")
print("=" * 70)

for i, marca in enumerate(marcas):
    print(i, repr(marca))
    lista.append(marca)

print()
print("Quantidade de marcas:", len(marcas))
print(lista)
```
